# Gráfico 6: precio de la vivienda frente a inflación (H2)

**Qué muestra:** si la línea de la inflación se mueve en la misma dirección que la del crecimiento del precio de la vivienda. Es asociación, no causalidad.

**Datos:** `merge_mediana_mundial.csv` (carpeta `data/Mergeados`).

## 1 Librerías y funciones propias (src/utils/archivos.py)

```python
import sys
sys.path.append("..")
from utils.archivos import cargar_dataset, guardar_grafico

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

## 2 Cargo los datos

```python
datos_graf = cargar_dataset("merge_mediana_mundial.csv")
datos_graf
```

## 3 Dibujo y guardo el gráfico

```python
plt.style.use("seaborn-v0_8")
plt.figure(figsize=(12, 6))

sns.lineplot(data=datos_graf, x="anio", y="crec._precio_(%)", label="Precio vivienda", color="red")
sns.lineplot(data=datos_graf, x="anio", y="inflacion_(%)", label="Inflación", color="blue", linestyle=":")

plt.yticks(np.arange(0, 11, 1))
plt.title("Precio de la vivienda e inflación: mediana mundial (25 países, 2016-2024)")
plt.xlabel("Año")
plt.ylabel("Crecimiento anual (%)")
plt.legend()
guardar_grafico("06_vivienda_vs_inflacion.png")
plt.show()
```
