```python
import pycountry
```

```python
# FUNCION PARA HOMOGENEIZAR TODOS LOS CODIGOS DE PAÍS Y PODER MERGEAR LOS 3 DATA SET (IA)
def codigo_a_iso3(codigo):
    pais = pycountry.countries.get(numeric=f"{codigo:03d}")
    if pais is None:
        return None
    return pais.alpha_3
```

```python
def codigos_a_iso3(codigos):
    resultado = []
    for codigo in codigos:
        pais = pycountry.countries.get(numeric=f"{codigo:03d}")
        if pais is None:
            resultado.append(None)
        else:
            resultado.append(pais.alpha_3)
    return resultado
```
