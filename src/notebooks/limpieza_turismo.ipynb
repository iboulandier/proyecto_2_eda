```python
print(data_turistas_pernoctan["partner_area_label"].unique())
print(data_turistas_pernoctan["unit"].unique())
print(data_turistas_pernoctan["partner_area_label"].nunique())
print(data_turistas_pernoctan["reporter_area_label"].nunique())
```

```python
data_turistas_pernoctan = data_turistas_pernoctan.drop(columns = ["indicator_code", "indicator_label", "indicator_previous_code",
    "partner_area_code", "partner_area_label",
    "flag", "flag_label", "unit", "notes"])
print(data_turistas_pernoctan)
```

```python
print(data_turistas_pernoctan.shape)
```

```python
data_turistas_pernoctan["value"] = (data_turistas_pernoctan["value"] * 1000).astype(int)
data_turistas_pernoctan
```

```python
data_turistas_pernoctan  = data_turistas_pernoctan.rename(columns={"reporter_area_code": "codigo_onu",
    "reporter_area_label": "pais",
    "year": "anio", "value": "numero_turistas"})
data_turistas_pernoctan
```
