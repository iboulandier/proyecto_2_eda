# Gráfico 1: evolución de la mediana mundial (vista completa)

Este gráfico muestra la mediana anual de las cuatro variables (precio de la vivienda, turismo, PIB per cápita e inflación) en los 25 países con serie completa 2016-2024, con los valores extremos del turismo a la vista. Se coge la mediana y no la media por la desviación en la media provocada por la crisis del coronavirus y la gran cantidad de outliers

**Datos:** `merge_mediana_mundial.csv` (mediana por año de los 25 países con datos completos 2016-2024).

## 1 Librerías

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

## 2 Cargo los datos desde carpeta "Mergeados"

```python
datos_grafica = pd.read_csv("../data/Mergeados/merge_mediana_mundial.csv")
datos_grafica
```

## 3 Dibujo y guardo el gráfico

```python
plt.style.use("seaborn-v0_8")
plt.figure(figsize=(12, 6))

sns.lineplot(data=datos_grafica, x="anio", y="crec._precio_(%)", label="Precio vivienda", color="red")
sns.lineplot(data=datos_grafica, x="anio", y="crec._turistas_(%)", label="Turistas", color="green")
sns.lineplot(data=datos_grafica, x="anio", y="crecim_pib_(%)", label="PIB per cápita", color="blue")
sns.lineplot(data=datos_grafica, x="anio", y="inflacion_(%)", label="Inflación", color="purple", linestyle=":")

plt.yticks(np.arange(-75, 151, 25))
plt.title("Evolución de la mediana mundial (25 países, 2016-2024)")
plt.xlabel("Año")
plt.ylabel("Crecimiento anual (%)")
plt.legend()
plt.savefig("01_evolucion_mediana_mundial.png", bbox_inches="tight")
plt.show()
```
