# Gráfico 2: evolución de la mediana mundial (con zoom)

Este gráfico muestra el mismo gráfico que el 1, con el eje Y limitado de -5 a 20 para ver mejor cómo se mueven precio, PIB e inflación. El turismo se sale del marco en 2020 (-74 %) y 2022 (+136 %).

**Datos:** `merge_mediana_mundial.csv` (mediana por año de los 25 países con datos completos 2016-2024).

## 1 Librerías

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

## 2 Cargo los datos

```python
datos_grafica_zoom = pd.read_csv("../data/Mergeados/merge_mediana_mundial.csv")
datos_grafica_zoom
```

## 3 Dibujo y guardo el gráfico

```python
plt.style.use("seaborn-v0_8")
plt.figure(figsize=(12, 6))

sns.lineplot(data=datos_grafica_zoom, x="anio", y="crec._precio_(%)", label="Precio vivienda", color="red")
sns.lineplot(data=datos_grafica_zoom, x="anio", y="crec._turistas_(%)", label="Turistas", color="green")
sns.lineplot(data=datos_grafica_zoom, x="anio", y="crecim_pib_(%)", label="PIB per cápita", color="blue")
sns.lineplot(data=datos_grafica_zoom, x="anio", y="inflacion_(%)", label="Inflación", color="purple", linestyle=":")

plt.ylim(-5, 20)
plt.yticks(np.arange(-5, 21, 2.5))
plt.title("Evolución de la mediana mundial (zoom: eje Y de -5 a 20)")
plt.xlabel("Año")
plt.ylabel("Crecimiento anual (%)")
plt.legend()
plt.savefig("02_evolucion_mediana_mundial_zoom.png", bbox_inches="tight")
plt.show()
```
